```python
import pandas as pd
import pickle as pkl
```

```python
with open("data/test.pkl", "rb") as file:
    test = pkl.load(file)

with open("data/train.pkl", "rb") as file:
    train = pkl.load(file)   
```

```python
X = pd.concat([train, test])
```

```python
Z = X.drop_duplicates()
```

```python
Z.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
train, test = train_test_split(Z, stratify=Z["Status"], test_size=0.1)
```

```python
with open("data/train.pkl", "wb") as file:
    pkl.dump(train, file)
```

```python
with open("data/test.pkl", "wb") as file:
    pkl.dump(test, file)
```
