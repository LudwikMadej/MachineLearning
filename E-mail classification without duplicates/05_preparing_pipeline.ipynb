```python
word_counts = [0 for i in range(df.shape[0])]
counter = 0
for index, row in df.iterrows():
    x = pd.Series(row["Message"])
    x = x[x != ""]
    x = x.value_counts().to_dict()

    table = np.zeros(len(importants_words), dtype=np.int16)

    for index in range(len(importants_words)):
        count = x.get(importants_words[index], 0)
        table[index] = count
    
    word_counts[counter] = table
    counter += 1
```
